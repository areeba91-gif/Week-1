```python
import pandas as pd

df = pd.read_csv("teaching_prices_clean.csv")
df.head()
```

```python
print("Shape:", df.shape)
print("Columns:", list(df.columns))
print("Distinct items:", df["item"].nunique())
print("Distinct dates:", df["date"].nunique())
print("Distinct cities:", df["city"].nunique())
print("First date:", df["date"].min(), " Last date:", df["date"].max())
```

```python
first = df[df["date"] == df["date"].min()].groupby("item")["price"].mean()
last = df[df["date"] == df["date"].max()].groupby("item")["price"].mean()

change = pd.DataFrame({"first_price": first, "last_price": last})
change["percent_change"] = ((last - first) / first * 100).round(1)
change
```

```python
change.sort_values("percent_change", ascending=False)
```
